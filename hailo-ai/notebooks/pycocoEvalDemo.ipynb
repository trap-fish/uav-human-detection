```python
%matplotlib inline
import matplotlib.pyplot as plt
from pycocotools.coco import COCO
from pycocotools.cocoeval import COCOeval
import numpy as np
import pylab

import json
pylab.rcParams['figure.figsize'] = (10.0, 8.0)
```

```python
annType = ['segm','bbox','keypoints']
annType = annType[1]      #specify type here
prefix = 'person_keypoints' if annType=='keypoints' else 'instances'
print('Running demo for *%s* results.'%(annType))
```

```python
#initialize COCO ground truth api
dataDir='./'
dataType='val2017'
# annFile = f"annotations/{prefix}_{dataType}.json"
annFile = '../visdrone/annotations_VisDroneHumans_val.json'
cocoGt=COCO(annFile)
```

```python
#initialize COCO detections api
#resFile=f"{dataDir}/results/{prefix}_{dataType}_fake{annType}100_results.json"
resFile="../visdrone/detections_VisDrone_val_quant_optlvl2_lbox3.json"
# resFile = '../visdrone/0000001_03999_d_0000007_VisDrone_quant_dets.json'
cocoDt=cocoGt.loadRes(resFile)
```

```python
imgIds=sorted(cocoGt.getImgIds())
imgIds = imgIds[0:]
#imgId = imgIds[np.random.randint(100)]

catIDs = [1,2] # 81 for all IDs
useCats = 0
maxDets = [1, 10, 100]
iou = [0.2, 0.5, 0.75]
```

```python
# running evaluation
cocoEval = COCOeval(cocoGt,cocoDt,annType)
cocoEval.params.imgIds  = imgIds
cocoEval.params.catIds = catIDs
cocoEval.params.maxDets = maxDets
cocoEval.params.useCats = useCats
#cocoEval.params.iouThrs = iou  # For IoU = 0.5
cocoEval.evaluate()
cocoEval.accumulate()
cocoEval.summarize()
```
